```python
import math
import os
from os import path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset_dir = '/kaggle/input/nasa-cmaps/CMaps'
output_dir = '/kaggle/working/data'
datasets = ['FD001', 'FD002', 'FD003', 'FD004']
```

```python
columns = ['unit_number', 'time_in_cycles', 'operational_setting_1', 'operational_setting_2', 'operational_setting_3']
sensor_columns = [f'sensor_measurement_{i}' for i in range(1, 27)]
columns += sensor_columns
```

```python
train_fd_dfs = []
test_fd_dfs = []

for i, dataset in enumerate(datasets):
    train_file = path.join(dataset_dir, f'train_{dataset}.txt')
    test_file = path.join(dataset_dir, f'test_{dataset}.txt')

    train_data = pd.read_csv(train_file, sep=' ', header=None, names=columns)
    test_data = pd.read_csv(test_file, sep=' ', header=None, names=columns)

    train_fd_dfs.append(train_data)
    test_fd_dfs.append(test_data)
```

```python
train_df = pd.concat(train_fd_dfs, ignore_index=True)
test_df = pd.concat(test_fd_dfs, ignore_index=True)
```

```python
def add_rul_to_training_data(train_df):
    # Step 1: Find the maximum life cycle for each engine unit.
    max_cycles_df = train_df.groupby('unit_number')['time_in_cycles'].max().reset_index()
    max_cycles_df.columns = ['unit_number', 'max_cycles']

    # Step 2: Merge the max_cycles information back into the main dataframe.
    train_df = pd.merge(train_df, max_cycles_df, on='unit_number', how='left')

    # Step 3: Calculate RUL = max_cycles - current_time_cycle.
    train_df['RUL'] = train_df['max_cycles'] - train_df['time_in_cycles']

    # Step 4: Drop the intermediate 'max_cycles' column.
    train_df.drop(columns=['max_cycles'], inplace=True)
    
    return train_df
```

```python
for train_fd_df in train_fd_dfs:
    train_fd_df = add_rul_to_training_data(train_fd_df)
```

```python
for train_fd_df in train_fd_dfs:
    train_fd_df.drop_duplicates(inplace=True)
```

```python
if not path.exists(output_dir):
    os.makedirs(output_dir)

for i in range(len(datasets)):
    dataset = datasets[i]
    train_fd_df = train_fd_dfs[i]
    test_fd_df = test_fd_dfs[i]

    if path.exists(path.join(output_dir, f'train_{dataset}_after_eda.csv')):
        os.remove(path.join(output_dir, f'train_{dataset}_after_eda.csv'))

    train_fd_df.to_csv(path.join(output_dir, f'train_{dataset}_after_eda.csv'), index=False)

    if path.exists(path.join(output_dir, f'test_{dataset}_after_eda.csv')):
        os.remove(path.join(output_dir, f'test_{dataset}_after_eda.csv'))

    test_fd_df.to_csv(path.join(output_dir, f'test_{dataset}_after_eda.csv'), index=False)
```
